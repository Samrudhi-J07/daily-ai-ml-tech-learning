```python
import gradio as gr
from transformers import pipeline
```

This loads a pretrained sentiment analysis model from Hugging Face.

It can identify whether text expresses:
- POSITIVE 😊
- NEGATIVE 😞

```python
model = pipeline("sentiment-analysis")
```

This function:
- Takes user text.
- Sends it to the AI model.
- Predicts sentiment.
- Returns the prediction and confidence score.

```python
def analyze(text):
    result = model(text)[0]

    return result["label"], round(result["score"], 3)
```

```python
app = gr.Interface(
    fn=analyze,
    inputs=gr.Textbox(label="Enter Your Text"),
    outputs=[
        gr.Textbox(label="Sentiment"),
        gr.Number(label="Confidence")
    ],
    title="AI Sentiment Analyzer",
    description="Analyze text using Artificial Intelligence"
)

app.launch(share=True)
```
